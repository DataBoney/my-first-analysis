## What is Google Colab?

Google Colab, also known as "Collaboratory," is a free, web-based environment hosted by Google that allows you to create, edit, and share documents containing:

- Live code
- Visualizations
- Explanatory text
- Equations

Using Google Colab, you can easily run code, document your process, and collaborate on projects directly in the cloud.

## Why Use Google Colab?

- **Interactive Coding**: Write and run code in small chunks (cells) and see the output instantly.
- **Data Visualization**: Easily create charts and graphs to visualize your data.
- **Documentation**: Use Markdown to add headings, lists, links, and images to explain your code and make it easier to follow.
- **Sharing and Collaboration**: Share your notebooks with others for collaboration or showcase your work.

## Switching Between Cell Types

Sometimes you might need to change a cell from code to Markdown or vice versa.

- **Change Cell Type**:
  - Select the cell.
  - Go to the toolbar and choose "Code" or "Markdown" from the dropdown menu.
  - Alternatively, you can use keyboard shortcuts:
    - `Esc` to enter command mode (cell border turns blue).
    - Press `M` to convert to Markdown.
    - Press `Y` to convert to Code.


# Additional Tips

## Running Cells

You can run cells individually, in groups, or all at once. Select Cell in the menu, and you'll see three commands:

* 'Run All' will run all cells.
* 'Run All Above' will run all cells above the selected one.
* 'Run All Below' will run all cells below the selected one.



# Common Questions

## Why is my code not working?

- **Check the Order**: Remember that variables and functions are available in the order cells are run. If you get a `NameError`, make sure you've run the cell where the variable or function is defined.
- **Restarting the Kernel**: If things get messy, you can restart the notebook's kernel (its computational engine) by going to "Kernel" > "Restart". This will clear all variables and you can start fresh.

## How do I save my work?

- **Autosave**: Jupyter Notebook autosaves your work periodically.
- **Manual Save**: Click the save icon (💾) or press `Ctrl + S` to save manually.

## Clearing the Notebook and Restarting

If you want to clear all outputs and restart the Python interpreter (kernel), you can do the following:

1.  **Restart Runtime**: Go to `Runtime` > `Restart runtime`. This will clear all variables and reset the kernel, but the cell outputs will remain.
2.  **Restart and Run All**: Go to `Runtime` > `Restart and run all`. This will clear all outputs and then run all cells from top to bottom.
3.  **Clear All Outputs**: Go to `Edit` > `Clear all outputs`. This will remove all cell outputs but keep the kernel state as is.
